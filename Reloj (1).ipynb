```python
import pandas as pd
import numpy as np
df = pd.read_excel("Ikinci_El_Saat_Veri_Seti_Guncel.xlsx")
print(df.shape) #verinin boyutu
uniquevalues = df.nunique()
print(uniquevalues) # for one-hot encoding
df.drop(columns=["Kondisyon", "Kutu_Sertifika"], inplace=True)
```

```python
columns = ['Marka', 'Mekanizma', 'Malzeme', 'Cam_Tipi', 'Kordon_Tipi', 'Kadran_Rengi']
df_encoded = pd.get_dummies(df, columns=columns)
df_encoded.to_excel('encoded_saat_verileri.xlsx', index=False)
from google.colab import files
files.download('encoded_saat_verileri.xlsx')
```

```python
from sklearn.preprocessing import MinMaxScaler
numeric_columns = ['Kasa_Capi_mm', 'Su_Gecirmezlik_m', 'Tahmini_Fiyat_TL']
scaler = MinMaxScaler()
df_encoded[numeric_columns] = scaler.fit_transform(df_encoded[numeric_columns])
df_encoded
```

```python
user_request = {
    'Marka': 'Seiko',
    'Kadran_Rengi': 'Beyaz',
    'Kordon_Tipi': 'Çok Parçalı Çelik',
    'Mekanizma': 'Otomatik',
    'Malzeme': 'Seramik',
    'Cam_Tipi': 'Akrilik',
    'Kasa_Capi_mm': 40,
    'Tahmini_Fiyat_TL': 25000,
    'Su_Gecirmezlik_m': 50
}
```

```python
vector = pd.DataFrame(0, index=[0], columns=df_encoded.columns)
print(vector)
```

```python
for anahtar, deger in user_request.items():
    if anahtar in numeric_columns:
      vector.at[0, anahtar] = deger
    else:
        sutun_adi = f"{anahtar}_{deger}"
        if sutun_adi in vector.columns:
            vector.at[0, sutun_adi] = 1
```

```python
vector[numeric_columns] = scaler.transform(vector[numeric_columns])
print(vector)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
space = df_encoded.to_numpy()
similarity_score = cosine_similarity(vector, space).flatten()
best_results = similarity_score.argsort()[::-1][:3]
best_results
```

```python
from google.colab import userdata
from google import genai
api_key = userdata.get('GEMINI_API_KEY')
client = genai.Client(api_key=api_key)
def saat_tavsiye_et(marka, kadran, kordon, mekanizma, malzeme, cam, kasa_capi, butce, su_gecirmezlik):
    user_request = {
    'Marka': marka,
    'Kadran_Rengi': kadran,
    'Kordon_Tipi': kordon,
    'Mekanizma': mekanizma,
    'Malzeme': malzeme,
    'Cam_Tipi': cam,
    'Kasa_Capi_mm': int(kasa_capi),
    'Tahmini_Fiyat_TL': float(butce),
    'Su_Gecirmezlik_m': int(su_gecirmezlik)
}
    vector = pd.DataFrame(0, index=[0], columns=df_encoded.columns)

    numeric_columns = ['Kasa_Capi_mm', 'Su_Gecirmezlik_m', 'Tahmini_Fiyat_TL']

    for anahtar, deger in user_request.items():
        if anahtar in numeric_columns:
            vector.at[0, anahtar] = deger
        else:
            sutun_adi = f"{anahtar}_{deger}"
            if sutun_adi in vector.columns:
                vector.at[0, sutun_adi] = 1
    vector[numeric_columns] = scaler.transform(vector[numeric_columns])
    min_butce = butce * 0.50
    max_butce = butce * 1.50
    uygun_indeksler = df[(df['Tahmini_Fiyat_TL'] >= min_butce) & (df['Tahmini_Fiyat_TL'] <= max_butce)].index
    if len(uygun_indeksler) == 0:
        return pd.DataFrame(columns=['Marka', 'Kadran_Rengi', 'Kordon_Tipi', 'Tahmini_Fiyat_TL'])

    filtrelenmis_uzay = df_encoded.loc[uygun_indeksler].to_numpy()
    filtrelenmis_orijinal = df.loc[uygun_indeksler]

    similarity_score = cosine_similarity(vector, filtrelenmis_uzay).flatten()
    best_indices = similarity_score.argsort()[::-1][:3]

    tavsiyeler = filtrelenmis_orijinal.iloc[best_indices][['Marka', 'Kadran_Rengi', 'Kordon_Tipi', 'Tahmini_Fiyat_TL']]

    saat_detaylari = tavsiyeler.to_string()
    prompt = f"""
    Sen profesyonel bir ikinci el saat eksperi ve stil danışmanısın.
    Kullanıcının bütçesi: {butce} TL.
    Kullanıcıya sistemin önerdiği en iyi 3 saat şunlar:
    {saat_detaylari}

    Bu saatlerin neden bu kullanıcıya uygun olduğunu, aralarındaki stil farklarını
    ve hangisini neden seçmesi gerektiğini samimi, şık ve akıcı bir dille (Türkçe) açıkla.
    """

    try:
        response = client.models.generate_content(
            model='gemini-3.6-flash',
            contents=prompt,
        )
        llm_yorumu = response.text
    except Exception as e:
        llm_yorumu = f"Yapay zeka yorumu yüklenirken bir hata oluştu: {str(e)}"

    return tavsiyeler, llm_yorumu
```

```python
demo = gr.Interface(
    fn=saat_tavsiye_et,
    inputs=[
        gr.Dropdown(choices=list(df['Marka'].unique()), label="Marka"),
        gr.Dropdown(choices=list(df['Kadran_Rengi'].unique()), label="Kadran Rengi"),
        gr.Dropdown(choices=list(df['Kordon_Tipi'].unique()), label="Kordon Tipi"),
        gr.Dropdown(choices=list(df['Mekanizma'].unique()), label="Mekanizma"),
        gr.Dropdown(choices=list(df['Malzeme'].unique()), label="Malzeme"),
        gr.Dropdown(choices=list(df['Cam_Tipi'].unique()), label="Cam Tipi"),
        gr.Number(label="Kasa Çapı (mm)", value=40),
        gr.Number(label="Bütçe (TL)", value=25000),
        gr.Number(label="Su Geçirmezlik (m)", value=50)
    ],
    outputs=[
        gr.Dataframe(label="Sizin İçin Seçtiğimiz En İyi 3 Saat"),
        gr.Textbox(label="Watchexpert AI - Stil ve Uzman Yorumu", lines=6)
    ],
    title="⌚ Akıllı İkinci El Saat Öneri Sistemi",
    description="Kriterlerini seç, yapay zeka vektör uzayında tarayıp bütçene ve zevkine en uygun saatleri getirsin!"
)

# Arayüzü çalıştır
demo.launch(debug=True)
```
